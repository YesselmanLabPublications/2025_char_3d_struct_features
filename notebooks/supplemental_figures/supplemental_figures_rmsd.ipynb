# Supplemental Figure RMSD

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from dms_3d_features.plotting import (
    plot_pop_avg_from_row,
    publication_style_ax,
    plot_violins_w_percent,
    plot_motif_boxplot_stripplot,
)
```

```python
FARFAR_PATH = "../../farfar-models"
```

```python
df_train = pd.read_json(f"{FARFAR_PATH}/summary/ag_pairs_train_summary.json")
```

```python
df_train["mean_rmsd_with_constraint"] = df_train["with_constraint"].apply(
    lambda x: x["top100_models"]["mean_rmsd"] if isinstance(x, dict) else None
)

df_train["mean_rmsd_without_constraint"] = df_train["no_constraint"].apply(
    lambda x: x["top100_models"]["mean_rmsd"] if isinstance(x, dict) else None
)
```

```python
kink_turns = [
    "CGACGAAG_CAGAG",
    "CGAUGAAA_UGGAG",
    "CGUUGAAA_UGGAG",
    "UGGGAUU_ACGAAG"
]

data_outliers = [
    "CGAC_GAG",
    "CGAUGAAA_UGGAG",
    "GGGAGC_GGGAGC",
    "UGGC_GAG",
    "AAGCAG_UCAGGU",
    "AAU_AGU"
]

# -----------------------------------
# SORT by WITH constraint (ascending)
# -----------------------------------
df_sorted = df_train.sort_values(
    by="mean_rmsd_without_constraint",

)
motif_order = df_sorted["motif"].tolist()
# -----------------------------------
# FIGURE with shared Y-axis
# -----------------------------------
fig, (ax1, ax2) = plt.subplots(
    ncols=2,
    figsize=(12, 12),
    sharey=True,
    gridspec_kw={"wspace": 0},
    dpi = 600
)

# -----------------------------------
# COLOR PALETTE (no category ordering)
# -----------------------------------
#palette = sns.color_palette("tab20", n_colors=df_sorted["category"].nunique())

# -----------------------------------
# LEFT PLOT → WITH CONSTRAINT
# -----------------------------------
sns.barplot(
    data=df_sorted,
    x="mean_rmsd_without_constraint",
    y="motif", 
    dodge=False,
    ax=ax1,
    edgecolor="black", 
    linewidth=0.8
)

ax1.set_xlabel("Mean RMSD without constraints")
ax1.set_ylabel("Motif")

df_sorted["mean_rmsd_diff"] =  df_sorted["mean_rmsd_without_constraint"] - df_sorted["mean_rmsd_with_constraint"]
# -----------------------------------
# RIGHT PLOT → DIFFERENCE (same as before)
# -----------------------------------
sns.barplot(
    data=df_sorted,
    x="mean_rmsd_diff",
    y="motif",
    dodge=False,
    ax=ax2,
    edgecolor="black", 
    linewidth=0.8
)

ax2.set_xlabel("Mean RMSD difference (no constraints - with constraint)")
ax2.set_ylabel("")  # remove duplicate label

# -----------------------------------
# REMOVE duplicate legends (keep only one)
# -----------------------------------
legend = ax2.get_legend()
if legend is not None:
    legend.remove()

# -----------------------------------
# MODIFY Y-TICK LABELS (INLINE SYMBOLS)
# -----------------------------------
new_labels = []

for motif in motif_order:

    if motif in kink_turns and motif in data_outliers:
        label = r"$\mathbf{*†}$ " + motif
    elif motif in kink_turns:
        label = r"$\mathbf{*}$ " + motif
    elif motif in data_outliers:
        label = r"$\mathbf{†}$ " + motif
    else:
        label = motif

    new_labels.append(label)

ax1.set_yticklabels(new_labels)

# -----------------------------------
# REFERENCE LINE (only on diff plot)
# -----------------------------------
ax2.axvline(0, linestyle="--", linewidth=0.5, color="black")
ax2.set_xlim(-2, 2)
ax2.tick_params(axis='y', left=False, labelleft=False)

# -----------------------------------
# FINAL LAYOUT
# -----------------------------------
publication_style_ax(ax1)
publication_style_ax(ax2)
plt.show()
```
